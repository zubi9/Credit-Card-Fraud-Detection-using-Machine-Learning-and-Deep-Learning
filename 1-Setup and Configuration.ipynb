````python
import gc

def reset_memory():
    # Delete all variables.
    for var in gc.get_objects():
        try:
            del var
        except:
            pass

    # Run the garbage collector.
    gc.collect()

    print("Memory has been reset")
````

## 1.5 Define Paths

````python
import os

DATA_DIR =  Path("data/raw")
# List the contents of the data/raw directory
print(f"Contents of {DATA_DIR}:")
for item in os.listdir(DATA_DIR):
    print(item)
````

````python

# Paths


# DATA_DIR = Path("data/raw")

TRAIN_TRANSACTION = DATA_DIR / "train_transaction.csv"
TRAIN_IDENTITY = DATA_DIR / "train_identity.csv"

TEST_TRANSACTION = DATA_DIR / "test_transaction.csv"
TEST_IDENTITY = DATA_DIR / "test_identity.csv"

SUBMISSION = DATA_DIR / "sample_submission.csv"
````

---

## 1.6 Load Data

````python

# Load Training Data


train_transaction = pd.read_csv(TRAIN_TRANSACTION)

train_identity = pd.read_csv(TRAIN_IDENTITY)

print("Transaction Shape :", train_transaction.shape)
print("Identity Shape    :", train_identity.shape)
````

## 1.7 Merge Datasets


````python

# Merge Transaction and Identity


df = train_transaction.merge(train_identity,
    how="left",
    on="TransactionID"
)

print("Merged Shape :", df.shape)
````

## 1.12 Missing Values Overview

We won't impute anything yet.

Instead, we'll quantify missingness.

````python
missing = (
    df.isnull()
      .sum()
      .sort_values(ascending=False)
      .to_frame("Missing")
)

missing["Percentage"] = (
    missing["Missing"] / len(df)
) * 100

missing.head(20)
````
