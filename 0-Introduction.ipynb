# Credit Card Fraud Detection using Machine Learning and Deep Learning

## Problem Statement

Financial institutions process millions of online payment transactions every day. While the vast majority of these transactions are legitimate, a small percentage are fraudulent and can lead to significant financial losses, operational costs, and reduced customer trust. Detecting fraudulent transactions is particularly challenging because fraudulent behavior continuously evolves, fraud cases are extremely rare compared to legitimate transactions, and transaction data often contain missing values, noisy information, and complex relationships between customers, devices, and payment methods.

Traditional rule-based fraud detection systems are limited in their ability to identify previously unseen fraud patterns. Machine learning provides a data-driven approach that can learn complex patterns from historical transaction data and detect suspicious activities in real time. However, building an effective fraud detection system requires much more than selecting a classification algorithm. It involves data acquisition, preprocessing, exploratory data analysis, feature engineering, handling class imbalance, model development, evaluation, explainability, and deployment considerations.

The objective of this project is to develop an end-to-end fraud detection pipeline capable of accurately distinguishing fraudulent transactions from legitimate ones using both classical machine learning algorithms and a deep neural network. The developed models will be evaluated using metrics suitable for highly imbalanced datasets, such as ROC-AUC, Precision-Recall AUC, F1-score, Precision, Recall, and confusion matrices, ensuring that the system minimizes financial losses while reducing false alarms.

# Project Description

In this project, we develop a complete Machine Learning Engineering pipeline for credit card fraud detection using the **IEEE-CIS Fraud Detection** dataset, released by the IEEE Computational Intelligence Society in collaboration with Vesta Corporation. The dataset represents real-world e-commerce transactions and contains over **590,000** labeled transactions with hundreds of transaction and identity-related features distributed across multiple tables. The data include payment card information, product details, email domains, device characteristics, browser information, anonymized numerical variables, and temporal attributes, providing a realistic environment for fraud detection research.

The project follows an industry-oriented workflow beginning with data acquisition and validation, followed by data cleaning, exploratory data analysis, feature engineering, preprocessing, and handling of missing values and class imbalance. Multiple supervised machine learning models—including Logistic Regression, Random Forest, XGBoost, LightGBM, and CatBoost—will be trained and compared against a Fully Connected Neural Network implemented using TensorFlow/Keras. Model performance will be assessed using appropriate evaluation metrics for imbalanced classification problems, and explainability techniques will be incorporated to understand the contribution of important features. The resulting pipeline demonstrates best practices for developing robust and interpretable fraud detection systems suitable for real-world financial applications.

I also suggest one improvement over a typical Kaggle notebook: we should not perform a random train/test split. The dataset has a temporal component (TransactionDT), and the original competition separated training and testing chronologically. Using a time-aware split for our validation will better simulate deployment and reduce the risk of temporal leakage.
