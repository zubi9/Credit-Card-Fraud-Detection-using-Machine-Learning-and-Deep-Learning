# 16.1 Project Summary


## Project Summary

The objective of this project was to develop an end-to-end machine learning system for credit card fraud detection using the IEEE-CIS Fraud Detection dataset. Unlike traditional machine learning demonstrations that primarily focus on model training, this project followed a complete machine learning engineering workflow covering the entire model development lifecycle.

The workflow included:

- Data acquisition and understanding
- Exploratory Data Analysis (EDA)
- Data cleaning and preprocessing
- Feature engineering
- Handling class imbalance
- Development of multiple machine learning models
- Development of a deep neural network baseline
- Threshold optimization
- Model explainability using SHAP
- Final evaluation on an unseen hold-out test set
- Model persistence and inference pipeline

The resulting notebook provides a reproducible and deployment-oriented fraud detection pipeline suitable for real-world machine learning applications.


# 16.2 Model Comparison Summary


## Model Performance Summary

Several supervised learning algorithms were evaluated, including:

- Logistic Regression
- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost
- Deep Neural Network

Each model was evaluated using the same validation strategy and performance metrics to ensure a fair comparison.

Since fraud detection is a highly imbalanced classification problem, model selection was based primarily on Precision-Recall AUC (PR-AUC), supported by ROC-AUC, Recall, Precision, and F1-score rather than overall Accuracy.


Choosing metrics beyond accuracy is essential because a classifier can achieve very high accuracy by predicting the majority class while failing to detect fraud. Precision, recall, F1-score, and PR-AUC provide a more meaningful assessment for imbalanced datasets. ([vizly.dev][1])

# 16.3 Best Model


## Selected Model

Among all evaluated models, LightGBM achieved the strongest overall performance.

The model demonstrated:

- Highest Precision-Recall AUC
- Strong ROC-AUC
- Good balance between Precision and Recall
- Fast training time
- Excellent scalability to large datasets
- Robust generalization on the hold-out test set

For these reasons, LightGBM was selected as the final production model.


# 16.4 Explainability


## Model Explainability

To improve model transparency and interpretability, SHAP (SHapley Additive exPlanations) was applied to the final LightGBM model.

The explainability analysis included:

- Global feature importance
- SHAP summary (beeswarm) plots
- Feature dependence analysis
- Local explanations using waterfall plots
- Individual transaction-level explanations

These analyses provide insight into the model's decision-making process and improve trust, debugging, and stakeholder communication.


Explainability is increasingly recognized as an important engineering practice for trustworthy ML systems, particularly in high-impact domains such as financial fraud detection. ([arXiv][2])

# 16.5 Limitations


## Limitations

Although the proposed solution achieved strong predictive performance, several limitations remain.

1. Hyperparameter optimization was intentionally deferred in order to first establish a complete end-to-end machine learning pipeline.

2. The evaluation considered only supervised learning methods. Semi-supervised and unsupervised anomaly detection techniques were outside the scope of this project.

3. Feature engineering was based on the provided dataset. Additional domain-specific behavioral features may further improve detection performance.

4. The notebook evaluated historical transaction data only and did not simulate streaming or real-time inference.

5. Probability calibration was evaluated but not further optimized using techniques such as Platt Scaling or Isotonic Regression.

# 16.7 Key Takeaways


## Key Takeaways

This project demonstrates that successful fraud detection extends beyond selecting a high-performing algorithm. A production-ready machine learning solution requires a structured engineering workflow that includes data preparation, robust validation, model comparison, explainability, reproducibility, and deployment considerations.

Following this process, LightGBM emerged as the most effective model for the IEEE-CIS Fraud Detection dataset, delivering the best balance between predictive performance, computational efficiency, and interpretability.

The resulting pipeline provides a strong foundation for real-world fraud detection systems and can be further extended with automated hyperparameter optimization, continuous monitoring, and production deployment.
